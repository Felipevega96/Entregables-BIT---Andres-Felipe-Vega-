Conclusiones: 

1. Las observaciones están dominadas por pequeños mamíferos.

De los 833 registros válidos, 421 corresponden a ardilla roja (50,5 %) y 278 a erizo (33,4 %); es decir, el 84 % de las observaciones son de pequeños mamíferos, frente a solo 71 de lince (8,5 %) y 63 de bisonte europeo (7,6 %). Esto indica que los resultados sobre lince y bisonte se basan en muestras pequeñas y son menos robustos. Además, un mayor número de registros no implica necesariamente mayor abundancia en la naturaleza: puede reflejar un mayor esfuerzo de muestreo o una mayor facilidad para detectar estas especies.

2. Las especies grandes tienen una distribución más restringida en el área de estudio.

La tabla cruzada país × especie muestra que el bisonte europeo solo se registró en Polonia (38), República Checa (16) y Eslovaquia (9), y que el lince no se observó en Alemania. En cambio, la ardilla roja y el erizo aparecen en los seis países. Esto sugiere que las especies de mayor tamaño ocupan un rango geográfico más limitado dentro del área de estudio, por lo que las comparaciones entre países para estas especies no pueden incluir a todos los países.

3. No hay evidencia suficiente de diferencias de tamaño del lince entre países.

La longitud promedio del lince varía entre 79,0 cm en Polonia (19 registros) y 87,4 cm en Eslovaquia (26 registros), una diferencia de 8,4 cm. Sin embargo, la desviación estándar de la longitud de los linces es de aproximadamente 15,1 cm, mayor que esa diferencia, y algunos países tienen muy pocos registros (Austria, 5). Esto indica que la variación entre países es comparable a la variación natural entre individuos, por lo que no se puede afirmar que los linces de un país sean más grandes que los de otro.

4. La limpieza de datos era indispensable para obtener resultados confiables.

El dataset original tenía 1.011 registros, de los cuales se eliminaron 178 (17,6 %): 12 filas sin información del animal y 166 duplicados. Además, un solo registro erróneo (un lince de 171 kg) elevaba la desviación estándar del peso de los linces de 4,7 kg a 18,3 kg, casi cuatro veces más. Esto demuestra que, sin diagnóstico ni limpieza, los análisis habrían producido conclusiones distorsionadas, y evidencia la necesidad de un protocolo estandarizado de captura de datos entre los investigadores.
